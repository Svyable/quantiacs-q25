## Strategy (long-only × is_liquid)

Replace body with the promoted mechanism. Keep Quantiacs fields only; no hand-picked coins.


```python
def strategy(data):
    close = data.sel(field="close")
    is_liquid = data.sel(field="is_liquid")
    # Example placeholder: SMA trend (replace with promoted logic)
    sma_fast = qnta.sma(close, 15)
    sma_slow = qnta.sma(close, 34)
    weights = xr.where(sma_fast > sma_slow, 1, 0) * is_liquid
    if "time" in getattr(weights, "dims", ()):
        # Multipass: last-day slice
        weights = weights.isel(time=-1)
    return weights
```
